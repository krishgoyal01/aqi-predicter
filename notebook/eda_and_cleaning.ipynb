feature engineering

```python
city_day.head()
```

```python
city_day = city_day.sort_values(['City', 'Date']).reset_index(drop=True)

city_day.head()
```

```python
city_day['AQI_lag1'] = city_day.groupby('City')['AQI'].shift(1)   # yesterday's AQI
city_day['AQI_lag2'] = city_day.groupby('City')['AQI'].shift(2)   # 2 days ago
city_day['AQI_lag3'] = city_day.groupby('City')['AQI'].shift(3)   # 3 days ago
city_day['AQI_lag7'] = city_day.groupby('City')['AQI'].shift(7)   # same day last week
```

```python
city_day['AQI_roll3'] = city_day.groupby('City')['AQI'].transform(lambda x: x.shift(1).rolling(3).mean())
city_day['AQI_roll7'] = city_day.groupby('City')['AQI'].transform(lambda x: x.shift(1).rolling(7).mean())

city_day = city_day.dropna()
print(city_day.shape)
```

```python
city_day['Month'] = city_day['Date'].dt.month
city_day['DayOfWeek'] = city_day['Date'].dt.dayofweek
city_day['Year'] = city_day['Date'].dt.year

# Optional but valuable: season, since Indian pollution is strongly seasonal
def get_season(month):
    if month in [12, 1, 2]:
        return 'Winter'
    elif month in [3, 4, 5]:
        return 'Summer'
    elif month in [6, 7, 8, 9]:
        return 'Monsoon'
    else:
        return 'Post-Monsoon'

city_day['Season'] = city_day['Month'].apply(get_season)
```
